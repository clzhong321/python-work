## 实验2：函数、条件、循环与递归

### 任务1：`score_level(score)` 成绩等级函数

90~100 → 优秀；80~90 → 良好；60~80 → 及格；0~60 → 不及格；超出 0~100 抛出 `ValueError`。

```python
# 任务1：成绩等级
def score_level(score):
    if not isinstance(score, (int, float)):
        raise ValueError(f"成绩必须是数值, 收到 {type(score).__name__}")
    if score < 0 or score > 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for s in [95, 85, 70, 50]:
    print(f"{s:3} -> {score_level(s)}")

try:
    score_level(120)
except ValueError as e:
    print("异常:", e)
```

### 任务2：`for` 与 `while` 计算 1 到 100 的和

```python
# 任务2：1 到 100 求和
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
n = 1
while n <= 100:
    total_while += n
    n += 1

print("for   循环结果:", total_for)
print("while 循环结果:", total_while)
print("两者相等:", total_for == total_while)
```

### 任务3：递归版与迭代版阶乘

约定 `0! = 1`，输入为非负整数。

```python
# 任务3：阶乘
def factorial_recursive(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    print(f"{n:>2}!  递归={factorial_recursive(n):>8}  迭代={factorial_iterative(n):>8}")
```

### 任务4：`summary(values)` 返回最小值、最大值、平均值

接收非空数值列表，以元组返回，并通过元组解包接收结果。

```python
# 任务4：summary 函数
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return (min(values), max(values), sum(values) / len(values))

data = [3, 1, 4, 1, 5, 9, 2, 6]
mn, mx, avg = summary(data)   # 元组解包
print(f"最小值={mn}, 最大值={mx}, 平均值={avg:.2f}")
```

## 实验4：文件读写、异常处理与模块

在 Notebook 同级目录创建 `exp01_data` 文件夹保存 UTF-8 文本文件 `students.txt`。

### 任务1：查看工作目录、构造相对路径、创建数据目录

```python
# 任务1：创建数据目录
import os
print("工作目录:", os.getcwd())
DATA_DIR = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(DATA_DIR, exist_ok=True)
print("数据目录:", DATA_DIR)
```

### 任务2：写入数据并读取（跳过表头、用字符串方法解析）

```python
# 任务2：写入 students.txt
content = (
    "name,score\n"
    "Alice,85\n"
    "Bob,58\n"
    "Carol,92\n"
    "David,76\n"
    "Eva,60\n"
    "Frank,89\n"
    "Grace,abc\n"
    "Henry,105\n"
    "Incomplete\n"
)
file_path = os.path.join(DATA_DIR, "students.txt")
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)

# 读取并解析
with open(file_path, encoding="utf-8") as f:
    lines = f.read().splitlines()
print("表头:", lines[0])
print("数据行(跳过表头):")
for line in lines[1:]:
    print(" ", repr(line))
```

### 任务3：`parse_student(line)` 函数

将成绩转换为整数，检查：字段数量、姓名非空、成绩在 0~100 范围内。无效记录抛出 `ValueError`。

```python
# 任务3：parse_student 函数
def parse_student(line):
    fields = line.split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量错误, 期望 2, 收到 {len(fields)}")
    name, score_str = fields
    name = name.strip()
    if not name:
        raise ValueError("姓名为空")
    score = int(score_str)   # 非数字会抛 ValueError
    if score < 0 or score > 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    return {"name": name, "score": score}

# 测试
print(parse_student("Alice,85"))
try:
    parse_student("Grace,abc")
except ValueError as e:
    print("无效记录:", e)
```

### 任务4：按行处理文件，捕获异常并继续

记录行号与错误原因，有效记录保存为字典组成的列表。

```python
# 任务4：按行处理文件
valid_records = []
invalid_count = 0
with open(file_path, encoding="utf-8") as f:
    next(f)  # 跳过表头
    for line_no, line in enumerate(f, start=2):
        line = line.rstrip("\n")
        if not line.strip():
            continue
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_count += 1
            print(f"第 {line_no} 行无效: {line!r} -> {e}")

print(f"\n有效记录数: {len(valid_records)}")
print(f"无效记录数: {invalid_count}")
print("有效记录:", valid_records)
```

### 任务5：写报告 `report.txt`，分别演示 `w` 与 `a` 模式

- `w`：覆盖写，若文件已存在则先清空。
- `a`：追加写，在文件末尾追加，不影响已有内容。

```python
# 任务5：写入报告
valid_scores = [r["score"] for r in valid_records]
valid_count = len(valid_records)
avg = sum(valid_scores) / len(valid_scores) if valid_scores else 0
pass_rate = sum(1 for s in valid_scores if s >= 60) / len(valid_scores) if valid_scores else 0

report_path = os.path.join(DATA_DIR, "report.txt")
# w 模式: 覆盖写
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {valid_count}\n")
    f.write(f"无效记录数: {invalid_count}\n")
    f.write(f"平均分: {avg:.2f}\n")
    f.write(f"及格率: {pass_rate:.2%}\n")

print("=== w 模式写入后 ===")
with open(report_path, encoding="utf-8") as f:
    print(f.read())

# a 模式: 追加写
with open(report_path, "a", encoding="utf-8") as f:
    f.write("--- 追加说明 (a 模式) ---\n")
    f.write("w 模式会清空原文件后写入, a 模式在文件末尾追加\n")

print("=== a 模式追加后 ===")
with open(report_path, encoding="utf-8") as f:
    print(f.read())
```

## 实验5：类、继承与对象复制

### 任务1：定义 `Student` 类

- `__init__()` 初始化 `name` 和 `score` 属性
- 实现 `is_passed()` 和 `__str__()` 方法

```python
# 任务1：Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name!r}, score={self.score})"
```

### 任务2：为实验4 中每条有效记录创建 `Student` 对象

```python
# 任务2：创建 Student 对象
for record in valid_records:
    stu = Student(record["name"], record["score"])
    print(f"{stu} -> 及格: {stu.is_passed()}")
```

### 任务3：定义 `GraduateStudent` 继承自 `Student`

- 增加 `research_topic` 属性
- 通过 `super().__init__()` 初始化父类属性
- 重写 `__str__()`

```python
# 任务3：GraduateStudent 类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)        # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name!r}, "
                f"score={self.score}, topic={self.research_topic!r})")

gs = GraduateStudent("Alice", 85, "深度学习")
print(gs)
print("及格:", gs.is_passed())
```

### 任务4：直接赋值、`copy.copy()`、`copy.deepcopy()` 比较

构造嵌套列表对象 `{"name": "Alice", "scores": [85, 90]}`，每次比较都从**新建的原始对象**开始。

```python
# 任务4：对象复制比较
import copy

def make_obj():
    return {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值（引用同一对象）
obj1 = make_obj()
ref = obj1
ref["scores"].append(100)
print("1. 直接赋值:")
print("   原对象:", obj1)
print("   ref   :", ref, "  # 与原对象相同, 都被修改")

# 2. copy.copy() 浅拷贝（外层独立, 内层共享）
obj2 = make_obj()
shallow = copy.copy(obj2)
shallow["scores"].append(100)
print("2. copy.copy() 浅拷贝:")
print("   原对象:", obj2, "  # 内部列表共享, 被影响")

# 3. copy.deepcopy() 深拷贝（完全独立）
obj3 = make_obj()
deep = copy.deepcopy(obj3)
deep["scores"].append(100)
print("3. copy.deepcopy() 深拷贝:")
print("   原对象:", obj3, "  # 完全独立, 不受影响")
```
