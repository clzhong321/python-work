## 字符串的乐趣

```python
# 1) 回文判断：忽略空格、标点和大小写
def is_palindrome(s):
    """判断字符串是否为回文（忽略空格、标点与大小写）。"""
    cleaned = "".join(ch.lower() for ch in s if ch.isalnum())
    return cleaned == cleaned[::-1]


print(is_palindrome("level"), is_palindrome("kayak"), is_palindrome("pop"))
print(is_palindrome("rats live on no evil star"))
print(is_palindrome("Was it a car or a cat I saw?"))
print(is_palindrome("T A C O cat"), is_palindrome("tacocat"))
print(is_palindrome("hello"))
```

```python
# 2) 字母顺序判断：忽略空格
def is_abecedarian(s):
    """判断字符串的字母是否从左到右按字母顺序排列（忽略空格）。"""
    cleaned = s.replace(" ", "").lower()
    for i in range(len(cleaned) - 1):
        if cleaned[i] > cleaned[i + 1]:
            return False
    return True


print(is_abecedarian("adder"), is_abecedarian("beet"))
print(is_abecedarian("dog"), is_abecedarian("cat"))
print(is_abecedarian("abcd efgh xyz"), is_abecedarian("tacocat"))
```

```python
# 3) 删除元音
def remove_vowels(s):
    """删除字符串中的所有元音 a、e、i、o、u。"""
    vowels = "aeiou"
    return "".join(ch for ch in s if ch.lower() not in vowels)


print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
print(remove_vowels("xyz"))
```

## 列表的乐趣

```python
# 1) 反转列表：参数不是列表时抛出 TypeError
def list_reverse(lst):
    """返回反转后的新列表；参数不是列表时抛出 TypeError。"""
    if not isinstance(lst, list):
        raise TypeError("list_reverse 的参数必须是列表")
    return lst[::-1]


print(list_reverse([1, 2, 3]))
print(list_reverse(["a", "b", "c", "d"]))
try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print("TypeError:", e)
```

```python
# 2) 判断序列是否升序：只遍历一次
def is_sorted(seq):
    """仅遍历一次判断序列是否按升序排列。"""
    it = iter(seq)
    try:
        prev = next(it)
    except StopIteration:
        return True
    for cur in it:
        if prev > cur:
            return False
        prev = cur
    return True


print(is_sorted([1, 2, 3, 4]))
print(is_sorted([1, 3, 2]))
print(is_sorted([]))
print(is_sorted((5, 5, 6)))
```

`is_sorted` 只遍历序列一次：用迭代器逐个取出元素，只需记住前一个元素与前前一次比较，
时间复杂度 O(n)，避免了对序列做多次完整遍历。


## 作为向量的元组

```python
# 1) 向量数乘：返回等长元组，每个元素乘以标量 s
def vec_scalar_mult(t, s):
    """向量数乘：返回与 t 等长的新元组，每个元素乘以标量 s。"""
    if not isinstance(t, tuple):
        raise TypeError("参数 t 必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("参数 s 必须是整数或浮点数")
    for x in t:
        if not isinstance(x, (int, float)):
            raise TypeError("元组 t 的元素必须都是整数或浮点数")
    return tuple(x * s for x in t)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, 2.5), 2.0))
try:
    vec_scalar_mult([1, 2, 3], 2)
except TypeError as e:
    print("TypeError:", e)
try:
    vec_scalar_mult((1, 2), "x")
except TypeError as e:
    print("TypeError:", e)
```

```python
# 2) 向量内积
def vec_inner_product(t1, t2):
    """返回两个向量元组的内积（浮点数）。"""
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个元组的长度必须一致")
    total = 0.0
    for a, b in zip(t1, t2):
        total += a * b
    return float(total)


print(vec_inner_product((1, 2, 3), (4, 5, 6)))
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
try:
    vec_inner_product([1, 2], (1, 2))
except TypeError as e:
    print("TypeError:", e)
```

```python
# 3) 校验矩阵：外层是元组、每个内层是等长元组、元素均为数字
def check_valid_mx(arg):
    """判断 arg 是否为有效矩阵。"""
    if not isinstance(arg, tuple):
        return False
    n_cols = None
    for row in arg:
        if not isinstance(row, tuple):
            return False
        if n_cols is None:
            n_cols = len(row)
        elif len(row) != n_cols:
            return False
        for x in row:
            if not isinstance(x, (int, float)):
                return False
    return True


print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2), (3, 4, 5))))   # 内层长度不一致 -> False
print(check_valid_mx(((1, 2), [3, 4])))       # 内层不是元组 -> False
print(check_valid_mx(((1, 2), (3, "4"))))     # 元素不是数字 -> False
print(check_valid_mx([1, 2]))                 # 不是元组 -> False
```

```python
# 4) 矩阵与向量相乘
def mx_vec_mult(mx, v):
    """矩阵（元组的元组）与向量（元组）相乘，返回数值元组。"""
    if not check_valid_mx(mx):
        raise ValueError("mx 不是有效矩阵")
    if not isinstance(v, tuple):
        raise TypeError("v 必须是元组")
    if len(mx) > 0 and len(mx[0]) != len(v):
        raise ValueError("向量长度必须等于矩阵的列数")
    return tuple(vec_inner_product(row, v) for row in mx)


my_mx = ((1, 2), (3, 4), (5, 6))
v = (1, 1)
print("矩阵 x 向量 =", mx_vec_mult(my_mx, v))

try:
    mx_vec_mult(my_mx, (1, 1, 1))
except ValueError as e:
    print("ValueError:", e)
try:
    mx_vec_mult([[1, 2], [3, 4]], v)
except ValueError as e:
    print("ValueError:", e)
```

## 更多向量的乐趣

```python
class Vector:
    """用面向对象方式表示的向量。"""

    # 1) 初始化：维度为非负整数，元素为整数/浮点数
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elements is None:
            elements = (0,) * dim          # 只给维度时，默认生成全零向量
        elif not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("元素个数必须与维度一致")
        for x in elements:
            if not isinstance(x, (int, float)):
                raise TypeError("元素必须是整数或浮点数")
        self.dim = dim
        self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # 3) 比较运算符
    def _check(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与另一个 Vector 进行比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        self._check(other)
        return self.elements == other.elements

    def __ne__(self, other):
        self._check(other)
        return self.elements != other.elements

    def __lt__(self, other):
        self._check(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check(other)
        return self.elements >= other.elements

    # 4) 内积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot 的参数必须是 Vector")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法计算内积")
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

    # 5) 乘法运算符
    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致，无法逐元素相乘")
            return Vector(self.dim, tuple(a * b for a, b in zip(self.elements, other.elements)))
        raise TypeError("只能与标量或 Vector 相乘")

    def __rmul__(self, other):
        return self.__mul__(other)


print(Vector(3, (1, 2, 3)))
print(Vector(3))                                  # 默认全零向量
print(Vector(2, (2, 4)) == Vector(2, (2, 4)))
print(Vector(2, (2, 4)) < Vector(2, (2, 5)))
print(Vector(2, (2, 5)) > Vector(2, (2, 4)))
print("内积:", Vector(2, (2, 4)).dot(Vector(2, (4, 5))))
print("2*v  :", 2 * Vector(2, (2, 3)))
print("v*2  :", Vector(2, (2, 3)) * 2)
print("v*w  :", Vector(2, (2, 3)) * Vector(2, (4, 5)))

for bad_call in (
    lambda: Vector(2, (1, 2, 3)),
    lambda: Vector(-1),
    lambda: Vector(2, (2, 4)) < Vector(3, (1, 2, 3)),
    lambda: Vector(2, (2, 4)) * Vector(3, (1, 2, 3)),
):
    try:
        bad_call()
    except (TypeError, ValueError) as e:
        print(f"{type(e).__name__}:", e)
```

### 2) 维度和元素应是类属性还是实例属性？

选择**实例属性**：在 `__init__` 中用 `self.dim`、`self.elements` 定义。
理由：每个向量对象有各自的维度和元素，不同向量取值不同；若写成类属性，则所有实例共享同一份数据，
修改一个实例会影响到其他实例，不符合“每个向量独立存储自身数据”的语义。类属性适合存放所有实例共用的常量。

### 3) 比较规则

两个 Vector 的坐标完全相同则相等；否则按元素元组的字典序比较（先比第一个坐标，再比第二个，依此类推），
例如 `(2, 4) < (2, 5)`。比较维度不同的向量时抛出异常。
